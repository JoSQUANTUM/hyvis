# Example Video Scan along a predefined trajectory

This notebook demonstrates the video_scan module of hyvis using a toy function. There are three ways to create videos with hyvis:  

- volume_scan: The simplest method. It scans a 3D subspace by treating the third dimension as time.
- operator_scan: A very flexible method. Transforms the subspace by an affine linear operation in each timestep.
- trajectory_scan: Methods for scanning the landscape along a predefined trajectory.

In general the functions create a scan object much like the basic scans but instead of calling scan.show() to view one plot you must call scan.animate() to play the video. If the video is to be saved, e.g. in the form of a gif one can use the following:

```python
# ani = scan.animate(vmin=-1, vmax=1)

# writervideo = animation.FFMpegWriter(fps=60)
# ani.save("scan.gif", writer=writervideo)
```

```python
import numpy as np
from hyvis.dr_tools import AffineSubspace
from hyvis.video_scans import (
    volume_scan,
    operator_scan,
    trajectory_scan_stepwise_pca,
)


def testfunc(arg: np.ndarray) -> float:
    return np.prod(np.cos(arg))
```

## Volume Scans

This works the same as the standard scans only that the last dimension is treated as time. This means that when you scan a 3D subspace with this method each frame will be a slice in the first two directions and when the video is played it will move through the third.

```python
subspace = AffineSubspace(
    directions=np.array([[1, 1, 0], [1, -1, 0], [0, 0, 1]])
)
scan = volume_scan(
    func=testfunc, subspace=subspace, scope=np.pi, resolution=[50, 50, 50]
)
scan.animate(vmin=-1, vmax=1)
```

## Operator Scans

Here after each frame is scanned the subspace is transformed by an affine linear oparator of your choice. This means the center is shifted and the space transformed by a matrix, eg a rotation. The matrix needs to be invertible and if it is not unitary the space will find its own orthogonal basis so its best to also guarantee that.

```python
theta = 2 * np.pi / 200

R = np.array(
    [
        [np.cos(theta), -np.sin(theta), 0],
        [np.sin(theta), np.cos(theta), 0],
        [0, 0, 1],
    ]
)

subspace = AffineSubspace(directions=np.array([[1, 1, 0], [1, -1, 0]]))
scan = operator_scan(
    func=testfunc,
    subspace=subspace,
    scope=10,
    operator=R,
    shift=np.array([[0, 0, 2 * np.pi / 200]]),
    steps=200,
    resolution=100,
)
scan.animate(vmin=-1, vmax=1)
```

## Trajectory Scans

It works in the following way: For each point in the trajectory a 2D-scan will be performed to make up a frame of the video. The directions of that scan are the two most significant principal components of the orthogonal projection of the remaining trajectory onto the orthogonal complement of the gradient at that point of the previous point. In other words, it scans the directions that show the most of the remaining trajectory but without considering the direction of the next point as that is where you will go in the next frame anyways.

```python
from hyvis.video_scans import trajectory_scan_stepwise_pca
```

Here we define the trajectory. It is just a path in the landscape with some multidimensional curvature given by sin and cos.

Importantly, resolution defines the resolution of each frame and step_num defines the length of the trajectory and thus the number of frames. These are the most significant parameters for performance.

```python
first = 2 * np.pi * np.array([0, 0, 0, 0])
last = 2 * np.pi * np.array([1, 1, 0, 0])
shift1 = 1 * np.array([0, 1, 1, 0])
shift2 = 1 * np.array([1, 0, 0, 1])

resolution = 50
step_num = 100
step_space = np.linspace(0, 1, step_num)

trajectory = np.zeros([step_num, 4])
for step_id in range(step_num):
    ratio = step_space[step_id]
    trajectory[step_id, :] = (
        first * (1 - ratio)
        + last * ratio
        + np.sin(2 * np.pi * ratio) * shift1
        + np.sin(np.pi * ratio) * shift2
    )
```

Now performing the scan:

```python
scan1 = trajectory_scan_stepwise_pca(
    func=testfunc, trajectory=trajectory, resolution=resolution, pools=10
)
```

And finally, looking at it with the animate method. This can take some time. The center of the frame is the current point and the red line shows the remaining trajectory with some fade towards the end. Note that this is a completely arbirary trajectory through this landscape.

The first and the last frame are just scans according to the full PCA, because the other method cannot be apllied at endpoints.

```python
scan1.animate(
    xlabel="first scan direction",
    ylabel="second scan direction",
    title="trajectory scan",
    show_trajectory=True,
    trajectory_fade=True,
    vmin=-1,
    vmax=1,
)
```

You can try this with any costfunction that maps a vector of real numbers to a real number,
but be warned that especially with the videos your function will be called many times.
